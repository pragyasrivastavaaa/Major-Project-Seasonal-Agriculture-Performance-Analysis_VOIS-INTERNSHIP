```python
# Example: median imputation for numerical columns
# To review missingness before deciding whether to use this approach.

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:", df.isnull().sum().sum())
```

# 9. Outlier Analysis

Investigate potential outliers using statistical and visual methods.

**Task:** Do not automatically delete outliers. Decide whether an extreme value is:

* A data-entry problem,
* A legitimate observation,
* Or something requiring further investigation.



### Univariate: Season distribution (Pie Chart)

```python
plt.figure(figsize=(7, 7))
df['Season'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title('Distribution of Records by Season')
plt.ylabel('') # Hide the default 'Season' label on the y-axis
plt.tight_layout()
plt.show()
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```

```python
# Outlier Detection for Key Financial and Production Metrics

outlier_cols = ["Yield_Tonnes_Ha", "Profit_INR", "Water_Used_m3", "Fertilizer_kg_ha"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, col in enumerate(outlier_cols):
    sns.boxplot(data=df, y=col, ax=axes[i], color="#2b5c8f")
    axes[i].set_title(f"Outlier Profile: {col}", fontsize=12, fontweight="bold")
    axes[i].set_ylabel("")

plt.suptitle("Outlier Distribution Across Critical Variables", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# Student Analysis 4: Multi-Dimensional Performance Profiling across Key Crops (Radar Chart)
# 1. Select key performance metrics to compare across crops
radar_cols = ["Yield_Tonnes_Ha", "Profit_INR", "Water_Efficiency", "Fertilizer_Efficiency"]

# Ensure custom efficiency columns exist
if "Water_Efficiency" not in df.columns:
    df["Water_Efficiency"] = df["Production_Tonnes"] / df["Water_Used_m3"]
if "Fertilizer_Efficiency" not in df.columns:
    df["Fertilizer_Efficiency"] = df["Yield_Tonnes_Ha"] / df["Fertilizer_Used_kg"]

# 2. Aggregate average performance per crop
crop_profiles = df.groupby("Crop")[radar_cols].mean()

# 3. Min-Max normalize the metrics (scale 0 to 1) for fair multi-axis comparison
crop_profiles_norm = (crop_profiles - crop_profiles.min()) / (crop_profiles.max() - crop_profiles.min())

# 4. Select top representative crops to highlight on the radar plot
selected_crops = ["Sugarcane", "Chilli", "Rice", "Wheat"]
filtered_profiles = crop_profiles_norm.loc[selected_crops]

# 5. Radar Chart Setup
labels = ["Yield Volume", "Profitability", "Water Efficiency", "Fertilizer ROI"]
num_vars = len(labels)

angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # Complete the circle loop

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Clean corporate palette mapping
colors = ["#2b5c8f", "#d95f02", "#7570b3", "#e7298a"]

for (crop_name, row), color in zip(filtered_profiles.iterrows(), colors):
    values = row.values.tolist()
    values += values[:1]  # Complete the circle loop
    ax.plot(angles, values, linewidth=2, linestyle='solid', label=crop_name, color=color)
    ax.fill(angles, values, color=color, alpha=0.15)

# Formatting axes and style
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
plt.xticks(angles[:-1], labels, size=11, fontweight="bold")
ax.set_rlabel_position(0)
plt.yticks([0.25, 0.50, 0.75, 1.0], ["25%", "50%", "75%", "100%"], color="grey", size=9)
plt.ylim(0, 1)

# Title shifted higher and Legend shifted slightly below the title line
plt.title("Crop Multi-Attribute Performance Profile (Normalized Radar Plot)\n", fontsize=14, fontweight="bold", pad=35)
plt.legend(loc="upper right", bbox_to_anchor=(1.20, 0.95))

plt.tight_layout()
plt.show()
```
